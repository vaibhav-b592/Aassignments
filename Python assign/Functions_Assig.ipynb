```python
# 1.Write a Python function to find the maximum of three numbers.
def maximum(a,b,c):
    if a > b:
        if a > c:
            return a,"is largest number"
    elif b > c:
        return b, "is largest number"
    elif c > b:
        return c, "is largest number"

print(maximum(3,8,1))
```

```python
# 2. Write a Python function to sum all the numbers in a list. Sample List : (8, 2, 3, 0, 7) Expected Output : 20

def sum_of_list(*L):
    sum = 0
    for i in L:
        sum = sum + i
    
    return sum

print(sum_of_list(8, 2, 3, 0, 7))
```

```python
""" 3. Write a Python function to multiply all the numbers in a list.

Sample List : (8, 2, 3, -1, 7) Expected Output : -336"""

def multiply(*L):
    product = 1
    for i in L:
        product *= i
    
    return product

print(multiply(9, 6, 3, -2, 7))
```

```python
"""4. Write a Python function to reverse a string.

Sample String: "1234abcd"
Expected Output: "dcba4321"
"""

class StringOperation:
    def __init__(self, text):
        self.text = text

    def reverse_string(self):
        return self.text[::-1]


obj = StringOperation("1234abcd")

print(obj.reverse_string())
```

```python
# 5. Write a Python function to check whether a number falls within a given range.abs

def check_range(a, b, n):
    if n > a:
        if n < b:
            print("Number is in the range")
        else:
            print("Number is out of range")
    elif n < a:
        print("Number is out of range")
    elif n > b:
        print("Number is out of range")


check_range(10, 50, 25)
```

```python
# 6. Write a Python function that accepts a string and counts the number of upper and lower case letters. Sample String : 'The quick Brow Fox' Expected Output : No. of Upper case characters : 3 No. of Lower case Characters : 12abs
def count_upper_lower(s):
    u = 0
    l = 0

    for ch in s:
        if ch.isupper():
            u += 1
        if ch.islower():
            l += 1

    return u, l


u, l = count_upper_lower("Hello Python WORLD")

print("No. of Uppercase letters:", u)
print("No. of Lowercase letters:", l)
```

```python
# 7 Write a Python function that takes a list and returns a new list with distinct elements from the first list.

#Sample List : [1,2,3,3,3,3,4,5] Unique List : [1, 2, 3, 4, 5]

def unique_list(numbers):
    result = []

    for num in numbers:
        if num not in result:
            result.append(num)

    return result


sample_list = [1, 2, 3, 3, 3, 3, 4, 5]

print("Original List:", sample_list)
print("Unique List:", unique_list(sample_list))
```

```python
# 8. Write a Python function that checks whether a passed string is a palindrome or not.
def palindrome_string(s):
    rev = s[::-1]

    if s == rev:
        print("String is palindrome")
    else:
        print("String is not palindrome")

palindrome_string("nitin")
```

```python
# 10.Write a Python program to detect the number of local variables declared in a function.
def my_function():
    local_vars = {}

    x = 15
    local_vars['x'] = x

    y = 25
    local_vars['y'] = y

    total = x + y
    local_vars['total'] = total

    message = "Python Programming"
    local_vars['message'] = message

    print("Local Variables:", local_vars)
    print("Number of local variables:", len(local_vars))


my_function()
```

```python
# 12. Write a Python program to create a function that takes one argument, and that argument will be multiplied with an unknown given number.

#Sample Output: Double the number of 15 = 30 Triple the number of 15 = 45 Quadruple the number of 15 = 60 Quintuple the number 15 = 75

def multiply_number(n, multiplier):
    return n * multiplier


number = 15

print("Double the number of", number, "=", multiply_number(number, 2))
print("Triple the number of", number, "=", multiply_number(number, 3))
print("Quadruple the number of", number, "=", multiply_number(number, 4))
print("Quintuple the number of", number, "=", multiply_number(number, 5))
```

```python
""" 13.Write a Python program to sort a list of tuples using Lambda.

Original list of tuples: [('English', 88), ('Science', 90), ('Maths', 97), ('Social sciences', 82)] Sorting the List of Tuples: [('Social sciences', 82), ('English', 88), ('Science', 90), ('Maths', 97)]%%! """
subjects = [
    ('English', 88),
    ('Science', 90),
    ('Maths', 97),
    ('Social sciences', 82)
]

sorted_list = sorted(subjects, key=lambda x: x[1])

print("Original list of tuples:", subjects)
print("Sorting the List of Tuples:", sorted_list)
```

```python
# 14. Write a Python program to check whether a given string contains a capital letter, a lower case letter, a number and a minimum length using lambda.abs
check_string = lambda s: (
    any(ch.isupper() for ch in s)
    and any(ch.islower() for ch in s)
    and any(ch.isdigit() for ch in s)
    and len(s) >= 8
)

text = "Python123"

if check_string(text):
    print("String is valid")
else:
    print("String is invalid")
```

```python
# 15. 21. Write a Python program to find numbers divisible by nineteen or thirteen from a list of numbers using Lambda.

#    Orginal list: [19, 65, 57, 39, 152, 639, 121, 44, 90, 190] Numbers of the above list divisible by nineteen or thirteen: [19, 65, 57, 39, 152, 190] show ans

numbers = [19, 65, 57, 39, 152, 639, 121, 44, 90, 190]

result = list(filter(lambda x: x % 19 == 0 or x % 13 == 0, numbers))

print("Original list:", numbers)
print("Numbers divisible by 19 or 13:", result)
```
