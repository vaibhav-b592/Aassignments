```python
# Handle ZeroDivisionError

num = 0  # Try changing this to 2, 5, or 0
num_list = [1, 24, 5, 9, 3]

for i in num_list:
    try:
        print(i / num)

    except ZeroDivisionError as z:
        print(f"{z}: Try again with another number except 0")
```

```python
#2. Write a Python program that prompts the user to input an integer and raises
#a ValueError exception if the input is not a valid integer
user_input = "25"

try:
    num = int(user_input)
    print("Valid integer:", num)

except ValueError:
    raise ValueError("Invalid input! Please enter a valid integer.")
```

```python
#3. Write a Python program that opens a file and handles a FileNotFoundError
# an exception if the file does not exist.
try:
    with open("sample.txt", "r") as file:
        content = file.read()
        print(content)

except FileNotFoundError:
    print("Error: The file does not exist.")
```

```python
# #5. Write a Python program that opens a file and handles a PermissionError exception if there is a permission issue.

try:
    with open("sample.txt", "r") as file:
        content = file.read()
        print(content)

except PermissionError:
    print("Error: You do not have permission to access this file.")
```

```python
#6. Write a Python program that executes an operation on a list and handles an IndexError exception if the index is out of range.

my_list = [10, 20, 30, 40, 50]

try:
    index = 10
    print("Element:", my_list[index])

except IndexError:
    print("Error: List index is out of range.")
```

```python
#7. Write a Python program that executes division and handles an ArithmeticError exception if there is an arithmetic error.
def exception_handle_fun():
    try:
        num1 = 10
        num2 = 0
        result = num1 / num2
        print("Result:", result)

    except ArithmeticError as e:
        print(f"Arithmetic error: {e}")

exception_handle_fun()
```
