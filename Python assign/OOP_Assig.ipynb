```python
# 2.Write a Python program to create a person class. Include attributes like name, country and date of birth. Implement a method to determine the person's age.
from datetime import datetime

class Person:

    def __init__(self,name,country,dob):
        self.name = name
        self.country = country
        self.dob = dob

    def person_age(self):
        dob = datetime.strptime(self.dob, "%Y-%m-%d")

        today = datetime.today()

        age = today.year - dob.year

        if (today.month,today.day) < (dob.month, dob.day):
            age -= 1

        return age
    
person = Person('Bob','USA','2007-06-23')
print(person.person_age())
```

```python

class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


class BinarySearchTree:
    def __init__(self):
        self.root = None

    # Insert an element
    def insert(self, data):
        if self.root is None:
            self.root = Node(data)
        else:
            self._insert(self.root, data)

    def _insert(self, root, data):
        if data < root.data:
            if root.left is None:
                root.left = Node(data)
            else:
                self._insert(root.left, data)

        elif data > root.data:
            if root.right is None:
                root.right = Node(data)
            else:
                self._insert(root.right, data)

    # Search for an element
    def search(self, data):
        current = self.root

        while current is not None:
            if data == current.data:
                return True
            elif data < current.data:
                current = current.left
            else:
                current = current.right

        return False



bst = BinarySearchTree()

# Insert elements
bst.insert(50)
bst.insert(30)
bst.insert(70)
bst.insert(20)
bst.insert(40)
bst.insert(60)
bst.insert(80)

# Search elements
print(bst.search(40))
print(bst.search(90))
```

```python
# 6.Write a Python program to create a class representing a stack data structure. Include methods for pushing and popping elements.

class Stack:

    def __init__(self):
        self.stack = []

    def push(self, el):
        self.stack.append(el)
        print(el, "pushed")

    def pop(self):
        if len(self.stack) == 0:
            print("Stack is empty")
        else:
            print(self.stack.pop(), "is popped")

    def display(self):
        print(self.stack)


stack = Stack()

stack.push(10)
stack.push(25)
stack.push(60)

stack.display()

stack.pop()
stack.display()
```

```python
# 7. Write a Python program to create a class representing a linked list data structure. Include methods for displaying linked list data, inserting and deleting nodes.abs
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    # Insert at end
    def insert(self, data):
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            return

        temp = self.head
        while temp.next:
            temp = temp.next
        temp.next = new_node

    def delete(self, key):
        temp = self.head

        # if head itself holds the key
        if temp and temp.data == key:
            self.head = temp.next
            temp = None
            return

        prev = None
        while temp and temp.data != key:
            prev = temp
            temp = temp.next

        if temp is None:
            print("Value not found")
            return

        prev.next = temp.next
        temp = None

    def display(self):
        temp = self.head
        while temp:
            print(temp.data, end=" -> ")
            temp = temp.next
        print("None")


ll = LinkedList()

ll.insert(10)
ll.insert(20)
ll.insert(30)

ll.display()

ll.delete(20)
ll.display()
```

```python
#8.Write a Python class Inventory with attributes like item_id, item_name, stock_count, and price, and methods like add_item, update_item, and check_item_details.

class Inventory:

    def __init__(self):
        self.item_list = {}

    def add_item(self, item_id, item_name, stock_count, price):
        self.item_list[item_id] = {
            "name": item_name,
            "stock": stock_count,
            "price": price
        }
        print("Item added successfully")

    def update_item(self, item_id, stock_count=None, price=None):
        if item_id in self.item_list:
            if stock_count is not None:
                self.item_list[item_id]["stock"] = stock_count

            if price is not None:
                self.item_list[item_id]["price"] = price

            print("Item updated successfully")
        else:
            print("Item not found")

    def check_item_details(self, item_id):
        if item_id in self.item_list:
            item = self.item_list[item_id]
            print("Item ID:", item_id)
            print("Item name:", item["name"])
            print("Item stock:", item["stock"])
            print("Item price:", item["price"])
        else:
            print("Item not found")


i = Inventory()


i.add_item(101, "Laptop", 15, 55000)
i.add_item(102, "Mouse", 30, 500)
i.add_item(103, "Keyboard", 20, 1200)


i.check_item_details(101)


i.update_item(102, stock_count=45, price=650)


i.check_item_details(102)
```

```python
# 9.Write a Python class BankAccount with attributes like account_number, balance, date_of_opening and customer_name, and methods like deposit, withdraw, and check_balance.

class BankAccount:

    def __init__(self, accountNumber, name, balance, date_of_opening):
        self.accountNumber = accountNumber
        self.name = name
        self.balance = balance
        self.date_of_opening = date_of_opening

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print("Deposited:", amount)
            print("Available balance:", self.balance)
        else:
            print("Amount must be positive")

    def withdrawal(self, amount):
        if amount <= 0:
            print("Amount must be positive")
        elif amount <= self.balance:
            self.balance -= amount
            print("Withdrawn:", amount)
            print("Available balance:", self.balance)
        else:
            print("Insufficient balance")

    def check_balance(self):
        print("Available balance:", self.balance)


b = BankAccount('789012', 'Rahul', 25000, "2024-06-15")

b.deposit(3000)
b.withdrawal(2000)
b.check_balance()
```

```python
"""10. Write a Python class Restaurant with attributes like menu_items, book_table, and customer_orders, and methods like add_item_to_menu, book_tables, and customer_order
Perform the following tasks now:
Now add items to the menu. Make table reservations. Take customer orders. Print the menu. Print table reservations. Print customer orders. Note: Use dictionaries and lists to store the data."""


class Restaurant:

    def __init__(self):
        self.menu_items = {}
        self.booked_tables = []
        self.customer_orders = []

    # add food items to the menu
    def add_menu_item(self, name, price):
        self.menu_items[name] = price
        print(f"{name} added to menu.")

    # Reserve a table
    def reserve_table(self, table_number):
        self.booked_tables.append(table_number)
        print(f"Table {table_number} booked successfully.")

    # place a customer order
    def place_order(self, item):
        if item in self.menu_items:
            self.customer_orders.append(item)
            print(f"{item} order placed successfully.")
        else:
            print(f"{item} is not available on the menu.")

    # display the menu
    def show_menu(self):
        print("\n--- Restaurant Menu ---")
        for item, price in self.menu_items.items():
            print(f"{item}: Rs. {price}")

    # display reservations
    def show_bookings(self):
        print("\n--- Booked Tables ---")
        print(self.booked_tables)

    # display customer orders
    def show_orders(self):
        print("\n--- Customer Orders ---")
        print(self.customer_orders)


# create a restaurant object
restaurant = Restaurant()

# Add menu items
restaurant.add_menu_item("Pizza", 180)
restaurant.add_menu_item("Pasta", 150)
restaurant.add_menu_item("Sandwich", 80)

# book tables
restaurant.reserve_table(5)
restaurant.reserve_table(8)

# place orders
restaurant.place_order("Pizza")
restaurant.place_order("Pasta")

# display information
restaurant.show_menu()
restaurant.show_bookings()
restaurant.show_orders()
```

```python
"""11. Write a Python class Employee with attributes like emp_id, emp_name, emp_salary, and emp_department and methods like calculate_emp_salary, emp_assign_department, and print_employee_details.
Sample Employee Data: "ADAMS", "E7876", 50000, "ACCOUNTING" "JONES", "E7499", 45000, "RESEARCH" "MARTIN", "E7900", 50000, "SALES" "SMITH", "E7698", 55000, "OPERATIONS"
Use 'assign_department' method to change the department of an employee. Use 'print_employee_details' method to print the details of an employee. Use 'calculate_emp_salary' method takes two arguments: salary and hours_worked, which is the number of hours worked by the employee. If the number of hours worked is more than 50, the method computes overtime and adds it to the salary. Overtime is calculated as following formula: overtime = hours_worked - 50 Overtime amount = (overtime * (salary / 50))"""


class Employee:

    def __init__(self, emp_id, emp_name, emp_salary, emp_department):
        self.emp_id = emp_id
        self.emp_name = emp_name
        self.emp_salary = emp_salary
        self.emp_department = emp_department

    # Calculate salary including overtime
    def calculate_salary(self, hours_worked):
        regular_hours = 50
        overtime = max(0, hours_worked - regular_hours)

        overtime_amount = overtime * (self.emp_salary / regular_hours)
        total_salary = self.emp_salary + overtime_amount

        print("Regular Salary:", self.emp_salary)
        print("Overtime Amount:", overtime_amount)
        print("Total Salary:", total_salary)

    # Change employee department
    def assign_department(self, department):
        self.emp_department = department
        print("Department updated successfully.")

    # Display employee information
    def display_details(self):
        print("\n--- Employee Details ---")
        print("Employee ID:", self.emp_id)
        print("Employee Name:", self.emp_name)
        print("Employee Salary:", self.emp_salary)
        print("Employee Department:", self.emp_department)


# Create employee object
emp = Employee(205, "Rahul", 40000, "Data Analytics")

# Calculate salary for 60 hours
emp.calculate_salary(60)

# Update department
emp.assign_department("Data Science")

# Display employee details
emp.display_details()
```

```python
""" 12. A university wants to automate their admission process. Students are admitted based on the marks scored in the qualifying exam. A student is identified by student id, age and marks in qualifying exam. Data are valid, if:

Age is greater than 20 Marks is between 0 and 100 (both inclusive) A student qualifies for admission, if

Age and marks are valid and Marks is 65 or more Write a python program to represent the students seeking admission in the university."""


class AdmissionProcess:

    def __init__(self, student_id, age, marks):
        self.student_id = student_id
        self.age = age
        self.marks = marks

    def check_eligibility(self):
        print("Student ID:", self.student_id)
        print("Student Age:", self.age)
        print("Marks Obtained:", self.marks)

        if self.age > 20 and self.marks >= 65:
            print("Student is eligible for admission.")
        elif self.age <= 20:
            print("Age is not eligible.")
        else:
            print("Marks are below criteria.")

# Create student objects
student1 = AdmissionProcess(101, 22, 78)
student2 = AdmissionProcess(102, 19, 85)
student3 = AdmissionProcess(103, 23, 55)

# Check eligibility
student1.check_eligibility()
print()

student2.check_eligibility()
print()

student3.check_eligibility()
```

```python
""" 13.Write a program that uses datetime module within a class. Enter manufacturing date and expiry date of the product. The program must display the years, months and days that are left for expiry.
"""

from datetime import datetime


class Product:
    def __init__(self, manuf_date, exp_date):
        self.manuf_date = manuf_date
        self.exp_date = exp_date

    def calculate_expiry(self):
        manufacture = datetime.strptime(self.manuf_date, "%Y-%m-%d")
        expiry = datetime.strptime(self.exp_date, "%Y-%m-%d")

        difference = (expiry - manufacture).days

        years = difference // 365
        remaining = difference % 365

        months = remaining // 30
        days = remaining % 30

        return years, months, days

    def display_result(self):
        years, months, days = self.calculate_expiry()

        print("Manufacturing Date:", self.manuf_date)
        print("Expiry Date:", self.exp_date)
        print(f"Product expiry duration: {years} Years, {months} Months, {days} Days")


# Main program
if __name__ == "__main__":
    product = Product("2025-11-06", "2026-06-12")
    product.display_result()
```
