```python
# ### NP17. Normalize data using NumPy (Min-Max scaling).
# > 💡 **Hint:** `arr = df[cols].values` then `(arr - arr.min(axis=0)) / (arr.max(axis=0) - arr.min(axis=0))`.
# Use `axis=0` to operate column-wise.

import numpy as np
import pandas as pd

df = pd.DataFrame({
    "Age": [18, 25, 35, 50],
    "Salary": [20000, 30000, 50000, 80000]
})

# Extract values
arr = df[["Age", "Salary"]].values

print(arr)

# Min-Max Scaling
normalized = (arr - arr.min(axis=0)) / (arr.max(axis=0) - arr.min(axis=0))

# Convert back to DataFrame
normalized_df = pd.DataFrame(normalized,columns=["Age", "Salary"])

print(normalized_df)
```

```python
# ### NP18. Standardize data using Z-score with NumPy.
# > 💡 **Hint:** `z = (arr - arr.mean(axis=0)) / arr.std(axis=0)`.
# Wrap the result in a DataFrame using the original column names.

import numpy as np
import pandas as pd

df = pd.DataFrame({
    "Age": [18, 25, 35, 50],
    "Salary": [20000, 30000, 50000, 80000]
})

# Extract values
arr = df[["Age", "Salary"]].values

# Z-score normalization (standardization)
z = (arr - arr.mean(axis=0)) / arr.std(axis=0)

# Convert back to DataFrame
z_df = pd.DataFrame(z, columns=["Age", "Salary"])

print(z_df)
```

```python
# ### NP40. Apply rolling computation using NumPy.
# > 💡 **Hint:** Use a list comprehension: `[np.mean(arr[max(0,i-w+1):i+1]) for i in range(len(arr))]` 
# to manually compute a rolling mean with window `w`. Compare with `df["col"].rolling(w).mean()`.

# Sample DataFrame
df = pd.DataFrame({
    "Sales": [10, 20, 30, 40, 50, 60]
})

# Convert column to NumPy array
arr = df["Sales"].to_numpy()

# Window size
w = 3

# Rolling mean using NumPy
rolling_mean_np = [
    np.mean(arr[max(0, i - w + 1):i + 1])
    for i in range(len(arr))
]

print("Rolling Mean (NumPy):")
print(rolling_mean_np)

# Rolling mean using Pandas
df["Rolling_Mean_Pandas"] = df["Sales"].rolling(w).mean()

print("\nDataFrame with Pandas Rolling Mean:")
print(df)
```
